```python
import pandas as pd
import numpy as np
from datasets import load_dataset
```

```python
import tiktoken

def count_tokens(text: str, model_name: str = "gpt-4") -> int:
    """Returns the number of tokens in a text string for a given LLM."""
    try:
        encoding = tiktoken.encoding_for_model(model_name)
    except KeyError:
        print(f"Warning: Model {model_name} not found. Defaulting to cl100k_base encoding.")
        encoding = tiktoken.get_encoding("cl100k_base")
    return len(encoding.encode(text))

cost = 0
```

```python
ds1 = load_dataset("christinacdl/binary_hate_speech")

test = ds1['test'].to_pandas()

# find total number of tokens on test dataset. tokens from the text column
total_tokens = 0
for i in range(len(test)):
    total_tokens += count_tokens(test['text'][i]) + 200

price_per_1m = 2.5
total_price1 = (total_tokens/1_000_000) * price_per_1m

print(f"Total tokens: {total_tokens}\nTotal price: {total_price1} dollars")

cost += total_price1
```

```python
ds8 = load_dataset("Rami/multi-label-class-github-issues-text-classification")
test = ds8['test'].to_pandas()

allowed_categories = ["bug", "feature", "question", "won't fix", "docs"]

def clean_element(lst):
    final = []
    for elem in lst:
        clean = elem.split('->')[-1]
        final.append(clean)
    return final

test['labels'] = test['labels'].apply(clean_element)
test = test[test['labels'].apply(lambda cats: all(c in allowed_categories for c in cats))]
test = test[test['labels'].apply(len) > 0]

test.reset_index(drop=True, inplace=True)

total_tokens = 0
for i in range(len(test)):
    total_tokens += count_tokens(test['title'][i]) + 200

price_per_1m = 2.5
total_price8 = (total_tokens/1_000_000) * price_per_1m

print(f"Total tokens: {total_tokens}\nTotal price: {total_price8} dollars")

cost += total_price8
```
