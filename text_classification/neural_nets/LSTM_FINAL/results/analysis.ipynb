```python
import pandas as pd
```

```python
results_binary1 = pd.read_csv('lstm_binary1.csv')
results_binary2 = pd.read_csv('lstm_binary2.csv')
results_binary3 = pd.read_csv('lstm_binary3.csv')
results_multiclass1 = pd.read_csv('lstm_multiclass1.csv')
results_multiclass2 = pd.read_csv('lstm_multiclass2.csv')
results_multiclass3 = pd.read_csv('lstm_multiclass3.csv')
results_multilabel1 = pd.read_csv('lstm_multilabel1.csv')
results_multilabel2 = pd.read_csv('lstm_multilabel2.csv')
results_multilabel3 = pd.read_csv('lstm_multilabel3.csv')

results_list = [results_binary1, results_binary2, results_binary3,
           results_multiclass1, results_multiclass2, results_multiclass3,
           results_multilabel1, results_multilabel2, results_multilabel3]

names = ['binary1', 'binary2', 'binary3', 'multiclass1', 'multiclass2', 'multiclass3', 'multilabel1', 'multilabel2', 'multilabel3']
results_dict = dict(zip(names, results_list))
```

```python
def get_average(list1, list2, list3):
    average = []
    for i in range(len(list1)):
        average.append((list1[i] + list2[i] + list3[i]) / 3)
    return average
```

```python
# for result in results_list:
#     test_acc = result['test_acc'].mean()
#     test_precision = result['test_prec'].mean()
#     test_recall = result['test_rec'].mean()

#     max_mem_usage_train = result['max_memory_usage_train'].mean()
#     max_mem_usage_test = result['max_memory_usage_test'].mean()
#     max_vram_usage_train = result['max_vram_usage_train'].mean()
#     max_vram_usage_test = result['max_vram_usage_test'].mean()
#     total_time_train = result['total_time_train'].mean()
#     total_time_test = result['total_time_test'].mean()

#     test_f1 = result['test_f1'].mean()

#     print(f"Test accuracy: {test_acc}")
#     print(f"Test precision: {test_precision}")
#     print(f"Test recall: {test_recall}")
#     print(f"Test f1: {test_f1}")
#     print(f"Max memory usage train: {max_mem_usage_train}")
#     print(f"Max memory usage test: {max_mem_usage_test}")
#     print(f"Max vram usage train: {max_vram_usage_train}")
#     print(f"Max vram usage test: {max_vram_usage_test}")
#     print(f"Total time train: {total_time_train}")
#     print(f"Total time test: {total_time_test}")
#     print('\n')


# do the loop above but for result_dict
for key, value in results_dict.items():
    test_acc = value['test_acc'].mean()
    test_precision = value['test_prec'].mean()
    test_recall = value['test_rec'].mean()

    max_mem_usage_train = value['max_memory_usage_train'].mean()
    max_mem_usage_test = value['max_memory_usage_test'].mean()
    max_vram_usage_train = value['max_vram_usage_train'].mean()
    max_vram_usage_test = value['max_vram_usage_test'].mean()
    total_time_train = value['total_time_train'].mean()
    total_time_test = value['total_time_test'].mean()

    test_f1 = value['test_f1'].mean()

    print(f"Results for {key}")
    print(f"Test accuracy: {test_acc}")
    print(f"Test precision: {test_precision}")
    print(f"Test recall: {test_recall}")
    print(f"Test f1: {test_f1}")
    print(f"Max memory usage train: {max_mem_usage_train}")
    print(f"Max memory usage test: {max_mem_usage_test}")
    print(f"Max vram usage train: {max_vram_usage_train}")
    print(f"Max vram usage test: {max_vram_usage_test}")
    print(f"Total time train: {total_time_train}")
    print(f"Total time test: {total_time_test}")
    print('\n')
```
