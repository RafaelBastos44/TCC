```python
import pandas as pd
from datasets import load_dataset
```

```python
columns_to_keep = ['seed', 'batch_size', 'learning_rate', 'test_f1s']
```

```python
binary1 = pd.read_csv('bert_binary1.csv')[columns_to_keep]
binary2 = pd.read_csv('bert_binary2.csv')[columns_to_keep]
binary3 = pd.read_csv('bert_binary3.csv')[columns_to_keep]
multiclass1 = pd.read_csv('bert_multiclass1.csv')[columns_to_keep]
multiclass2 = pd.read_csv('bert_multiclass2.csv')[columns_to_keep]
multiclass3 = pd.read_csv('bert_multiclass3.csv')[columns_to_keep]
multilabel1 = pd.read_csv('bert_multilabel1.csv')[columns_to_keep]
multilabel2 = pd.read_csv('bert_multilabel2.csv')[columns_to_keep]
multilabel3 = pd.read_csv('bert_multilabel3.csv')[columns_to_keep]
```

```python
def get_best_params(df):
    df['avg_f1'] = df['test_f1s'].apply(lambda x: sum(eval(x)) / len(eval(x)))
    best_params = df.loc[df['avg_f1'].idxmax()]
    print(f"Batch size: {best_params['batch_size']}, "
          f"Learning rate: {best_params['learning_rate']}")
```

```python
print("Binary 1:")
get_best_params(binary1)
print("Binary 2:")
get_best_params(binary2)
print("Binary 3:")
get_best_params(binary3)
print("Multiclass 1:")
get_best_params(multiclass1)
print("Multiclass 2:")
get_best_params(multiclass2)
print("Multiclass 3:")
get_best_params(multiclass3)
print("Multilabel 1:")
get_best_params(multilabel1)
print("Multilabel 2:")
get_best_params(multilabel2)
print("Multilabel 3:")
get_best_params(multilabel3)
```
