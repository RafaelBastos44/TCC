```python
from datasets import load_dataset
import numpy as np

ds = load_dataset("TimSchopf/arxiv_categories", "default")

df_train = ds['train'].to_pandas()
df_test = ds['test'].to_pandas() 
df_val = ds['validation'].to_pandas()
```

```python
def clean_element(lst):
    final = []
    for elem in lst:
        clean = elem.split('->')[-1]
        final.append(clean)
    return final

df_train['categories'] = df_train['categories'].apply(clean_element)
df_test['categories'] = df_test['categories'].apply(clean_element)
df_val['categories'] = df_val['categories'].apply(clean_element)
```

```python
allowed_categories = ["cs.AI", "cs.CL", "stat.ML", "math.OC", "cs.LG"]

def filter_categories(lst):
    return [x for x in lst if x in allowed_categories]
```

```python
df_train['categories'] = df_train['categories'].apply(filter_categories)
df_test['categories'] = df_test['categories'].apply(filter_categories)
df_val['categories'] = df_val['categories'].apply(filter_categories)

df_train = df_train[df_train['categories'].apply(len) > 0]
df_test = df_test[df_test['categories'].apply(len) > 0]
df_val = df_val[df_val['categories'].apply(len) > 0]
```
